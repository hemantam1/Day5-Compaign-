```python
import requests
import json

params = {
  'api_key': '41CCD23CF84449F6B7F0CA75B4D1DC1B',
  'type': 'search',
  'amazon_domain': 'amazon.in', 
  'search_term': 'aloe vera juice' 
}

api_result = requests.get('https://api.rainforestapi.com/request', params)
data = api_result.json()

competitor_list = []

if 'search_results' in data:
    for product in data['search_results'][:10]: 
        info = {
            'ASIN': product.get('asin'),
            'Title': product.get('title')[:50] + "...", 
            'Price': product.get('price', {}).get('value', 'N/A'),
            'Rating': product.get('rating', 'N/A')
        }
        competitor_list.append(info)
    print("\n--- TOP 10 COMPETITORS FOUND ---")
    for comp in competitor_list:
        print(f"ASIN: {comp['ASIN']} | Price: ₹{comp['Price']} | Rating: {comp['Rating']}")
else:
    print("No data found kindly check for the appropriate keyword.")
```
